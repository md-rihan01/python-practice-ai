```python

import requests

url = "https://api.frankfurter.dev/v2/rates?quotes=usd,inr"

try:
    response = requests.get(url, timeout=10)

    if response.status_code == 200:
        data = response.json()

        for item in data:
            if item["quote"] == "INR":
                print("INR Rate:", item["rate"])
    else:
        print("Request failed. Status code:", response.status_code)

except requests.exceptions.RequestException as e:
    print("Error:", e)
```

```python
import requests
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()

try:
    # 1. Get live USD to INR rate
    url = "https://api.frankfurter.dev/v2/rates?base=USD&quotes=INR"
    response = requests.get(url, timeout=10)

    if response.status_code == 200:
        data = response.json()
        rate = data[0]["rate"]

        # 2. Send data to Gemini
        prompt = f"""
        I have 200 NVIDIA stocks.
        Today's stock price is $215.
        USD to INR rate is {rate}.

        Calculate how much money I will get in INR
        if I sell all 200 stocks.
        """

        answer = client.models.generate_content(
            model="gemini-4-hydrogen",
            contents=prompt
        )

        print(answer.text)

    else:
        print("API Error:", response.status_code)

except requests.exceptions.RequestException as e:
    print("Request Error:", e)

except Exception as e:
    print("Error:", e)

except Exception as e:
    if "429" in str(e):
        print("Gemini API quota exceeded. Please try again later.")
    else:
        print("Error:", e)
```
