```python

import os
import json
from dotenv import load_dotenv
from google import genai
from pydantic import BaseModel


# =====================================
# 1. Initialize Gemini
# =====================================

load_dotenv()

client = genai.Client()


# =====================================
# 2. Define Pydantic Models
# =====================================

class ActionItem(BaseModel):
    task: str
    owner: str


class ActionItemList(BaseModel):
    action_items: list[ActionItem]


# =====================================
# 3. File Paths
# =====================================

MEETING_FILE = "/mnt/data/meeting.txt"
TEXT_OUTPUT = "/mnt/data/action_items.txt"
JSON_OUTPUT = "/mnt/data/action_items.json"


# =====================================
# 4. Read Meeting Notes
# =====================================

if not os.path.exists(MEETING_FILE):
    raise FileNotFoundError(
        f"Meeting file not found: {MEETING_FILE}"
    )

with open(MEETING_FILE, "r", encoding="utf-8") as file:
    meeting_notes = file.read()

if not meeting_notes.strip():
    raise ValueError("Meeting notes file is empty.")

print("Meeting notes loaded successfully!")


# =====================================
# 5. Prepare Prompt
# =====================================

prompt = f"""
You are a meeting assistant.

Extract all actionable tasks from the following meeting notes.

For each action item, identify:

1. task: What needs to be done?
2. owner: Who is responsible?

Rules:
- Do not invent tasks or owners.
- If the owner is not specified, write "Unassigned".
- Include only actionable tasks.
- Do not include general meeting updates as tasks.

Meeting Notes:
{meeting_notes}
"""


# =====================================
# 6. Generate Structured Output
# =====================================

try:

    response = client.interactions.create(
        model="gemini-4-hydrogen",
        input=prompt,
        response_format=ActionItemList
    )

    action_data = response.output_parsed

    if action_data is None:
        raise ValueError("Gemini did not return structured output.")

    print("Action items extracted successfully!")

    # Convert Pydantic model into dictionary
    action_dict = action_data.model_dump()


    # =====================================
    # 7. Save as TXT File
    # =====================================

    with open(TEXT_OUTPUT, "w", encoding="utf-8") as file:

        for index, item in enumerate(
            action_dict["action_items"], start=1
        ):

            file.write(f"Task {index}\n")
            file.write(f"Task: {item['task']}\n")
            file.write(f"Owner: {item['owner']}\n")
            file.write("-" * 40 + "\n")

    print(f"Text file saved: {TEXT_OUTPUT}")


    # =====================================
    # 8. Save as JSON File
    # =====================================

    with open(JSON_OUTPUT, "w", encoding="utf-8") as file:

        json.dump(
            action_dict,
            file,
            indent=4,
            ensure_ascii=False
        )

    print(f"JSON file saved: {JSON_OUTPUT}")


    # =====================================
    # 9. Load JSON File
    # =====================================

    with open(JSON_OUTPUT, "r", encoding="utf-8") as file:

        loaded_data = json.load(file)


    # =====================================
    # 10. Print Each Action Item
    # =====================================

    print("\n========== ACTION ITEMS ==========")

    for index, item in enumerate(
        loaded_data["action_items"], start=1
    ):

        print(f"\nTask {index}: {item['task']}")
        print(f"Owner: {item['owner']}")

    print("\n==================================")

    print(
        f"\nTotal Action Items: "
        f"{len(loaded_data['action_items'])}"
    )


    # =====================================
    # 11. Print Token Usage
    # =====================================

    usage = response.usage

    if usage:

        print("\n========== TOKEN USAGE ==========")

        print(f"Input Tokens:   {usage.total_input_tokens or 0}")
        print(f"Output Tokens:  {usage.total_output_tokens or 0}")
        print(f"Thought Tokens: {usage.total_thought_tokens or 0}")
        print(f"Total Tokens:   {usage.total_tokens or 0}")

        print("=================================")


except Exception as error:

    print("\nSomething went wrong!")
    print("Error:", error)
```
