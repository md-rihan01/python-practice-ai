```python

from dotenv import load_dotenv
from google import genai

# Load API key from .env
load_dotenv()

# Create Gemini client
client = genai.Client()

# Store the latest interaction ID
previous_interaction_id = None


def chat(user_message):
    global previous_interaction_id

    try:
        # Prepare request
        request = {
            "model": "gemini-4-hydrogen",
            "input": user_message
        }

        # Continue previous conversation
        if previous_interaction_id:
            request["previous_interaction_id"] = previous_interaction_id

        # Send message to Gemini
        interaction = client.interactions.create(**request)

        # Save interaction ID for conversation history
        previous_interaction_id = interaction.id

        # Get AI response
        response = interaction.output_text or "No response generated."

        # Print AI response
        print("\nGemini:", response)

        # Print token usage
        usage = interaction.usage

        if usage:
            print("\n--- Token Usage ---")
            print(f"Input Tokens:   {usage.total_input_tokens or 0}")
            print(f"Output Tokens:  {usage.total_output_tokens or 0}")
            print(f"Thought Tokens: {usage.total_thought_tokens or 0}")
            print(f"Tool Tokens:    {usage.total_tool_use_tokens or 0}")
            print(f"Total Tokens:   {usage.total_tokens or 0}")
            print("-------------------")

        return response

    except Exception as error:
        print(f"\nError: {error}")
        return None


# Start chatbot
print("Gemini Chatbot Started!")
print("Type 'exit' to stop the chatbot.\n")

while True:
    user_message = input("You: ").strip()

    if user_message.lower() in ["exit", "quit", "bye"]:
        print("Chatbot stopped. Goodbye!")
        break

    if not user_message:
        print("Please enter a message.")
        continue

    chat(user_message)
```
