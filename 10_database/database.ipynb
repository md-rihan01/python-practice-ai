```python
import sqlite3

conn = sqlite3.connect("shop.db")

cursor = conn.cursor()


cursor.execute("""
    CREATE TABLE IF NOT EXISTS reviews (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        text TEXT NOT NULL,
        label TEXT NOT NULL,
        score REAL
    )
""")

conn.commit()


reviews = [
    ("This product is amazing!", "Positive", 0.95),
    ("Very bad experience.", "Negative", 0.10),
    ("Good quality and fast delivery.", "Positive", 0.90),
    ("Not worth the money.", "Negative", 0.20),
    ("Average product.", "Neutral", 0.50)
]

cursor.executemany("""
    INSERT INTO reviews (text, label, score)
    VALUES (?, ?, ?)
""", reviews)

conn.commit()

print("Reviews inserted successfully!")

cursor.execute("SELECT * FROM reviews")

data = cursor.fetchall()

for row in data:
    print(row)

cursor.execute("SELECT * FROM reviews WHERE label = ?", ("Positive",))

data = cursor.fetchall()

for row in data:
    print(row)
```
