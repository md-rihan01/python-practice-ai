```python
from dotenv import load_dotenv
load_dotenv()
from google import genai
def generate_recipe(ingradiants, cuisine, diet):
    prompt = f'''
    Genrate one food recipe using these ingradiants: {','.join(ingradiants)}
    Recipe should not be more then 100 words
    Cuisine: {cuisine}
    Diet: {diet}
    '''
    client = genai.Client()
    interaction = client.interactions.create(
        model="gemini-4-hydrogen",
        input=prompt
    )
    return interaction.output_text

recipe = generate_recipe(["Tomatoes", "Lentils", "Onion", "Jeera", "salt"], "Indian", "Vegan")
print(recipe)
```
