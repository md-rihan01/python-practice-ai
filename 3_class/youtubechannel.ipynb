```python
class YouTubeChannel:
    def __init__(self, name):
        self.name = name
        self.subscribes = 0


    def subscribe(self):
        self.subscribes += 1
        return f"Thans for subscribing! Total: {self.subscribes}"

    def unsubscribe(self):
        if self.subscribes > 0:
            self.subscribes -= 1
        return f"Total: {self.subscribes}"

    def show(self):
        return f"{self.name} has {self.subscribes} subscribers"

kidsaiworld = YouTubeChannel("kidsaiworld")
kidsaiworld.subscribe()
kidsaiworld.subscribe()
kidsaiworld.subscribe()
kidsaiworld.unsubscribe()
kidsaiworld.show()
```
