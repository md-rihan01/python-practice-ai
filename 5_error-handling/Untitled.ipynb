```python

def safe_chai_bill(cups, price_per_cup):
    try:
        cups = int(cups)
        price_per_cup = float(price_per_cup)

    except ValueError:
        return "Please enter numbers only."

    if cups < 1:
        return "Cups must be at least 1."

    total_bill = cups * price_per_cup

    return total_bill


# Test cases
print(safe_chai_bill("3", "15"))   # Normal text input
print(safe_chai_bill(2, 15))       # Normal numbers
print(safe_chai_bill("two", 15))   # Invalid input
print(safe_chai_bill(0, 15))       # Too few cups
```
