```python
students = {
    "Ararv": [85, 90, 78],
    "Priya": [72, 68, 75],
    "Rohan": [45, 52, 48],
    "Sneha": [95, 92, 98],
    "Manish": [60, 65, 70]
}
print("=== Student Averages ===")
for stu, marks in students.items():
    for mark in marks:
        avg = sum(marks)/len(marks)
    if avg >= 80:
        print(f"{stu}: {avg:.2f} - Topper")
    elif avg >= 60 and avg < 79:
        print(f"{stu}: {avg:.2f} - Pass")
    elif avg < 60:
        print(f"{stu}: {avg:.2f} - Nees Improvements")


        
```

```python
print("\n===== TOPPERS =====")

for stu, marks in students.items():
    avg = sum(marks) / len(marks)

    if avg >= 80:
        print(f"{stu}: {avg:.2f}")
```

```python
print("\n===== AVERAGE STUDENTS =====")

for stu, marks in students.items():
    avg = sum(marks) / len(marks)

    if 60 <= avg < 80:
        print(f"{stu}: {avg:.2f}")
```
